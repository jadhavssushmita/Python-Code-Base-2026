### 4. Importing the dataset

```python
df = pd.read_csv('zara.csv', sep = ';')
```

### 5. Cleaning Stage

```python
## update date time datatype
df['scraped_at'] = pd.to_datetime(df['scraped_at']).dt.date
```

```python
## Rename columns
df = df.rename(columns = {'terms':'category'})
```

```python
## Drop a column
df = df.drop(columns = ['Product ID'])
```

### 6. Exploratory data analysis & Insights

```python
# What are categories of clothes in merchandise
df['category'].unique()
```

```python
## What are unique type of Jackets?
df[df['category'] == 'jackets']['name']
```

```python
## What are different types of shoes?
df[df['category'] == 'shoes']['name']
```

```python
#
df.head(1)
```

```python
## How many section categories?
df.section.unique()
```

```python
df.columns
```

```python
## Section wise what is the average price, total sales and product count?
df.groupby(['section']).agg(
    avg_price = ('price','mean'),
    total_sales = ('Sales Volume', 'sum'),
    product_count = ('sku', 'count')
    
)
```

```python
## what is the average price, total sales and product count by section and category?
df.groupby(['section','category']).agg(
    avg_price = ('price','mean'),
    total_sales = ('Sales Volume', 'sum'),
    product_count = ('sku', 'count')
    
)
```

```python
## Which categories were promoted the most compared to others?
df[df['Promotion'] == 'Yes'].groupby(['category','Promotion'])['sku'].count().sort_values(ascending = False)
```

```python
## How many products were promoted in each section ?
df.groupby(['section','Promotion'])['sku'].count()
```

```python
## Which category brings high sales?
df.nlargest(1,'price')
```

```python
# Total number of products
df['sku'].nunique()
```

```python
#Total products and Total products per category
df.groupby('category')['sku'].count()
```

```python
df.head()
```

```python
#Top 2 expensive products in each category

# .loc is short for label location
# Input: Give me the value for column product name at Row 0
df.loc[0, 'name']
```

```python
df.head(1)
```

```python
# iloc 
# Give me what value is sitting on x row and y column
# row, column
df.iloc[0,2]
```

```python
# lambda input: expression
df['price_category'] = df['price'].apply(lambda p: "Expensive" if p > 100 else "Affordable")
```

```python
## Top 2 expensive products in each category
df.groupby('category').apply(lambda x:x.nlargest(2,'price'))[['category','section','name','price']]
```

```python
df.dtypes
```

```python
# Do men shop more than women in Zara?
df.groupby('section').agg(
    total_sales_volume = ('Sales Volume', 'sum'),
    total_products = ('sku', 'count')
)
```
