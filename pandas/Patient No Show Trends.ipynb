### What key factors drive patient no-shows, and how can we design targeted interventions to improve attendance?



> Question: How to identify "No Show" in dataset? <br>
Answer: Use the appointment status column and filter for "No Show" <br> <br>

```python
# How many appointments are no show
df[df['appointment_status'] == 'No Show']['patient_id'].count()
```

> Question: What is the overall baseline no show rate? <br>
Answer: Create binary variable to check no show and then calculate mean 


```python
df['no_show_binary'] =  df['appointment_status'].apply(lambda x:1 if x == 'No Show' else 0)
```

```python
df['no_show_binary'].mean() * 100
```

> ~ 11.55 % standard benchmark 

> <b> How much is missed consulting fee due to 'No Show' appointment?


```python
df[df['appointment_status'] == 'No Show']['consultation_fee'].sum()
```

> A reminder call before ahead of schedule to prevent Rs 7 lakh loss

> <b> Do patients with missed previous appointments are likely to have no show


```python
df['app_missed_risk_indicator'] = df['missed_previous_appointments'].apply(lambda x : 'Low' if x <= 2 else ('Medium' if x <= 6 else 'High'))
```

```python
df.groupby('app_missed_risk_indicator').agg(
    total = ('patient_id', 'count'),
    noshowpct = ('no_show_binary', 'mean') 
).sort_values(by = 'noshowpct', ascending = False)

```

> <b> What is the precentage of missed appointments within patients?


```python
df[df['missed_previous_appointments'] != 0]['patient_id'].count() / df['patient_id'].count() * 100
```

> 67% patients have missed atleast one appointment in the past
